```python
import os
import time
import subprocess
import win32gui
import win32con

def show_window(hwnd, nCmdShow):
    win32gui.ShowWindow(hwnd, nCmdShow)

def init_sap_process(sap_working_directory, sap_path):
    # Set the current directory to the SAP working directory
    os.chdir(sap_working_directory)

    # Start the SAP process
    sap_process = subprocess.Popen(sap_path)

    # Wait for the window to become available
    hwnd = None
    while hwnd is None:
        hwnd = win32gui.FindWindow(None, None)
        time.sleep(0.5)

    # Show the window
    show_window(hwnd, win32con.SW_SHOWNORMAL)

# Example usage
sap_working_directory = r'C:\Path\To\SapWorkingDirectory'
sap_path = r'C:\Path\To\SapExecutable.exe'
init_sap_process(sap_working_directory, sap_path)
```
